```python
# Criar a estrutura de pastas do pacote
# A primeira leishimmunity é a pasta raiz do projeto.
# A segunda leishimmunity é a pasta que conterá os arquivos .py do pacote (é a pasta principal do pacote Python).
!mkdir -p leishimmunity/leishimmunity
# Criar subspastas
# leishimmunity/tests: Pasta para armazenar testes automatizados do pacote.
# leishimmunity/data: Pasta para armazenar dados brutos (ex. arquivos baixados do GEO).
# leishimmunity/results: Pasta para armazenar resultados gerados (ex. gráficos, tabelas).
!mkdir -p leishimmunity/tests
!mkdir -p leishimmunity/data
!mkdir -p leishimmunity/results
# Criar arquivos vazios
!touch leishimmunity/leishimmunity/__init__.py # Arquivo que torna a pasta um pacote Python.
# Criar arquivos .py para os módulos
!touch leishimmunity/leishimmunity/datasets.py # Arquivo com funções para baixar e carregar dados
!touch leishimmunity/leishimmunity/preprocess.py # Arquivo com funções para pré-processar dados
!touch leishimmunity/leishimmunity/analysis.py # Arquivo com funções para análise de expressão gênica
!touch leishimmunity/leishimmunity/visualize.py # Arquivo com funções para gerar gráficos
!touch leishimmunity/leishimmunity/utils.py # Arquivo com funções auxiliares como download de arquivos
# Criar arquivos de configuração
!touch leishimmunity/setup.py # Arquivo para instalar o pacote localmente.
!touch leishimmunity/requirements.txt # Lista de dependências do pacote como bibliotecas Python
!touch leishimmunity/README.md # Arquivo de documentação do projeto (explicação)
# Criar arquivo de teste
!touch leishimmunity/tests/test_datasets.py # Arquivo para testes automatizados das funções do módulo datasets.py.

```

!: Indica que o comando será executado no terminal do L

mkdir: Comando para criar diretórios (pastas).

-p: Flag que permite criar pastas aninhadas (ou seja, pastas dentro de pastas) sem erros, mesmo que algumas pastas já existam.

touch: Comando para criar um arquivo vazio.

Todo pacote Python precisa ter um arquivo __init__.py (pode estar vazio ou conter código de inicialização).
